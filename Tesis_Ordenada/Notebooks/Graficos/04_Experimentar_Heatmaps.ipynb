## 1. Importaciones y Configuración

```python
# Importaciones.
import sys
from pathlib import Path
import pandas as pd
import numpy as np
from scipy.stats import spearmanr

# Agregar rutas.
Ruta_Base = Path.cwd().parent.parent
sys.path.append(str(Ruta_Base / "Codigo" / "Utilidades"))
sys.path.append(str(Ruta_Base / "Codigo" / "Visualizacion"))

from Configuracion import *
from Graficos_Heatmap import *

print("✓ Módulos importados correctamente")
```

## 2. Cargar Datos y Calcular Correlaciones

```python
# Cargar datos procesados.
Ruta_Generales = RUTA_DATA_PROCESADOS / "Generales.xlsx"

if Ruta_Generales.exists():
    Df_Generales = pd.read_excel(Ruta_Generales)
    print(f"✓ Generales cargado: {len(Df_Generales)} filas")
else:
    print("⚠️ Archivo Generales.xlsx no encontrado")
    # Crear datos de ejemplo.
    Df_Generales = pd.DataFrame({
        'Categoria_PASO_2023': np.random.choice(
            ORDEN_CATEGORIAS,
            1000
        ),
        'CO_Pro_Izq': np.random.randn(1000),
        'CO_Pro_Der': np.random.randn(1000),
        'CO_Con_Izq': np.random.randn(1000),
        'CO_Con_Der': np.random.randn(1000),
        'CT_Pro_Izq': np.random.randn(1000),
        'CT_Con_Der': np.random.randn(1000)
    })
    print("✓ Usando datos de ejemplo")
```

```python
# Calcular matriz de correlación de ejemplo.
Variables_Ejemplo = [
    'CO_Pro_Izq',
    'CO_Pro_Der',
    'CO_Con_Izq',
    'CO_Con_Der'
]

# Verificar que las variables existen.
Variables_Disponibles = [
    v for v in Variables_Ejemplo
    if v in Df_Generales.columns
]

if Variables_Disponibles:
    Matriz_Correlacion = Df_Generales[
        Variables_Disponibles
    ].corr(method='spearman')
    print(
        f"✓ Matriz de correlación calculada: "
        f"{Matriz_Correlacion.shape}"
    )
else:
    # Crear matriz de ejemplo.
    Matriz_Correlacion = pd.DataFrame(
        np.random.rand(4, 4),
        index=Variables_Ejemplo,
        columns=Variables_Ejemplo
    )
    print("✓ Usando matriz de correlación de ejemplo")
```
