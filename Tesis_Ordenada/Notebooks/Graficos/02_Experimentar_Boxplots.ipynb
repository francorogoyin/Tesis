## 2. Cargar Datos

```python
# Cargar datos procesados (ajusta la ruta según necesites).
Ruta_Generales = RUTA_DATA_PROCESADOS / "Generales.xlsx"
Ruta_Ballotage = RUTA_DATA_PROCESADOS / "Ballotage.xlsx"

# Si los archivos existen, cargarlos.
if Ruta_Generales.exists():
    Df_Generales = pd.read_excel(Ruta_Generales)
    print(f"✓ Generales cargado: {len(Df_Generales)} filas")
else:
    print("⚠️ Archivo Generales.xlsx no encontrado")
    # Crear datos de ejemplo.
    Df_Generales = pd.DataFrame({
        'Categoria_PASO_2023': np.random.choice(
            ORDEN_CATEGORIAS,
            1000
        ),
        'CO_Pro_Izq': np.random.randn(1000),
        'CO_Con_Der': np.random.randn(1000)
    })
    print("✓ Usando datos de ejemplo")

if Ruta_Ballotage.exists():
    Df_Ballotage = pd.read_excel(Ruta_Ballotage)
    print(f"✓ Ballotage cargado: {len(Df_Ballotage)} filas")
else:
    print("⚠️ Archivo Ballotage.xlsx no encontrado")
    Df_Ballotage = pd.DataFrame({
        'Categoria_PASO_2023': np.random.choice(
            ORDEN_CATEGORIAS,
            500
        ),
        'CO_Pro_Izq': np.random.randn(500),
        'CO_Con_Der': np.random.randn(500)
    })
    print("✓ Usando datos de ejemplo")
```

# EXPERIMENTA AQUÍ

## Boxplot Básico

**Cambia estos parámetros:**

```python
# ============================================================
# PARÁMETROS PARA EXPERIMENTAR
# ============================================================

# Variable a graficar (prueba con diferentes variables).
Variable = 'CO_Pro_Izq'  # Cambia por: 'CO_Con_Der', etc.

# Título personalizado (None para automático).
Titulo = None  # O escribe: 'Mi Título Personalizado'

# Mostrar valores atípicos (outliers).
Mostrar_Outliers = True  # Cambia a False para ocultar

# Escala del eje Y (None para automático).
Escala_Y = None  # O usa: (-2, 2) para rango fijo

# Tamaño de figura (ancho, alto) en pulgadas.
Tamano_Figura = (12, 6)  # Prueba: (14, 7), (10, 5), etc.

# ============================================================
# EJECUTAR GRÁFICO
# ============================================================

Boxplot_Por_Categoria(
    Df=Df_Generales,
    Variable=Variable,
    Titulo=Titulo,
    Escala_Y=Escala_Y,
    Mostrar_Outliers=Mostrar_Outliers,
    Tamano_Figura=Tamano_Figura,
    Ruta_Salida=None  # None para mostrar, o Path para guardar
)
```

## Boxplot con Puntos Individuales

**Cambia estos parámetros:**

```python
# ============================================================
# PARÁMETROS PARA EXPERIMENTAR
# ============================================================

# Variable a graficar.
Variable_Puntos = 'CO_Con_Der'

# Transparencia de puntos (0 = invisible, 1 = opaco).
Alpha_Puntos = 0.3  # Prueba: 0.1, 0.5, 0.7

# Tamaño de puntos.
Tamano_Puntos = 3  # Prueba: 2, 4, 5

# ============================================================
# EJECUTAR GRÁFICO
# ============================================================

Boxplot_Con_Puntos(
    Df=Df_Generales,
    Variable=Variable_Puntos,
    Alpha_Puntos=Alpha_Puntos,
    Tamano_Puntos=Tamano_Puntos,
    Ruta_Salida=None
)
```

## Boxplots Múltiples (Cuadrícula)

**Cambia estos parámetros:**

```python
# ============================================================
# PARÁMETROS PARA EXPERIMENTAR
# ============================================================

# Lista de variables a graficar.
Variables_Multiples = [
    'CO_Pro_Izq',
    'CO_Pro_Der',
    'CO_Con_Izq',
    'CO_Con_Der'
]

# Títulos personalizados (None para automático).
Titulos_Personalizados = [
    'CO Progresistas Izquierda',
    'CO Progresistas Derecha',
    'CO Conservadores Izquierda',
    'CO Conservadores Derecha'
]

# Configuración de cuadrícula.
Filas = 2
Columnas = 2

# ============================================================
# EJECUTAR GRÁFICO
# ============================================================

Boxplots_Multiples(
    Df=Df_Generales,
    Variables=Variables_Multiples,
    Titulos=Titulos_Personalizados,
    Filas=Filas,
    Columnas=Columnas,
    Ruta_Salida=None
)
```
